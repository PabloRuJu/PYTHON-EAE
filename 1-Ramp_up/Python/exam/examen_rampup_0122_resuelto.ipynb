Las cuotas hipotecarias se calculan utilizando la fórmula del interés frances. 

La formula del interés frances es la siguiente:
\begin{align*}
C=V \frac{(1+i)^n i}{(1+i)^n -1}
\end{align*}


    C = Cuota a pagar (mensual)
    V = Cantidad del préstamo hipotecario
    i = tipo de interés del período (mensual)
    n = número de cuotas (mensual)

### 1️⃣ Create a class called VideoGameCharacter with the following attributes:

lives (int, default = 3)

level (int)

power (float)

nickname (str)

It must also contain two methods:

level_up() → increases the level by 1 and returns the updated level.

change_nickname(new_nickname) → updates the nickname but does not return anything.



```python
class VideoGameCharacter:
    lives = 3

    def __init__(self, level, power, nickname): #instance attributes
        self.level = level
        self.power = power
        self.nickname = nickname

    def level_up(self):
        self.level += 1
        return self.level
    
    def change_nickname(self, new_nickname):
        self.nickname = new_nickname
```
