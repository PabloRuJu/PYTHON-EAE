## Exercise 1
* Create two numeric variables: an `int` and a `float`
* Check their types
* Add them in another new variable
* What type is the new variable?
* Delete the first two variables created

````python
# Create two numeric variables
my_int = 10          # int
my_float = 3.5       # float

# Check their types
print(type(my_int))     # <class 'int'>
print(type(my_float))   # <class 'float'>

# Add them in a new variable
result = my_int + my_float

# Check the type of the new variable
print(type(result))     # <class 'float'>
print("result")

# Delete the first two variables
del my_int
del my_float
````

## Exercise 2
Write a program to convert degrees to radians. You must use `input`. Remember that the conversion is done through

radians = degrees*(pi/180)

````python
import math

# Ask the user for degrees
degrees = float(input("Enter the angle in degrees: "))

# Convert to radians
radians = degrees * (math.pi / 180)

# Display the result
print("Radians:", radians)
````

## Exercise 7
