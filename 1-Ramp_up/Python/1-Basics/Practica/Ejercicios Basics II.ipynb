## Exercise 1
* Create two numeric variables: an `int` and a `float`
* Check their types
* Add them in another new variable
* What type is the new variable?
* Delete the first two variables created

````python
# Create two numeric variables
my_int = 10          # int is a whole number
my_float = 3.5       # float has decimals

# Check their types
print(type(my_int))     # <class 'int'>
print(type(my_float))   # <class 'float'>

# Add them in a new variable
result = my_int + my_float

# Check the type of the new variable
print(type(result))     # <class 'float'>
print("result")

# Delete the first two variables
del my_int
del my_float
````

## Exercise 2
Write a program to convert degrees to radians. You must use `input`. Remember that the conversion is done through

radians = degrees*(pi/180)

````python
import math #Imported library for math functions

# Ask the user for degrees
degrees = float(input("Enter the angle in degrees: "))

# Convert to radians
radians = degrees * (math.pi / 180)

# Display the result
print("Radians:", radians)
````

## Exercise 10
For the following string you are asked to print the following cases on screen

    "Early birds catch the worm"

1. Convert it all to uppercase
2. Convert it all to lowercase
3. Convert to uppercase only the initial of the words
4. Create a list by dividing it by its spaces
5. Replace commas `,` with semicolons `;`
6. Remove lowercase `a` letters

````python
text1 = "Early birds, catch the worm"

# 1. Convert all to uppercase
print(text1.upper())

# 2. Convert all to lowercase
print(text1.lower())

# 3. Uppercase only the initial of the words
print(text1.title())

# 4. Create a list dividing by spaces
print(text1.split(" "))

# 5. Replace commas with semicolons
print(text1.replace(",", ";"))

# 6. Remove lowercase 'a' letters
print(text1.replace("a", ""))
````

## Exercise 11
