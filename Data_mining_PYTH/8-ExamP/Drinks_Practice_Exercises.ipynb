## Part 5: Multiple Conditions (AND / OR)

These exercises practice filtering with multiple conditions.

### Exercise 19: Find African countries with more than 100 wine servings

Display countries that are in Africa AND have more than 100 wine servings.

**Hint:** Use `&` to combine conditions. Don't forget parentheses around each condition!

```python
# Your code here
drinks[(drinks['continent']== 'AF') & (drinks['wine_servings'] > 100)] # Double specific conditions
```

### Exercise 20: Find countries with high beer OR high wine consumption

Display countries where beer servings > 300 OR wine servings > 300.

**Hint:** Use `|` (pipe) for OR logic. Remember the parentheses!

```python
# Your code here
drinks[(drinks['beer_servings'] > 300) | (drinks['wine_servings'] > 300)] #Double specific condition
                                    # With 'Or'
```

### Exercise 21: Find Asian countries with total alcohol > 5 litres

Display Asian countries with more than 5 litres of total pure alcohol consumption.

**Hint:** Combine `continent == 'AS'` AND `total_litres_of_pure_alcohol > 5`.

```python
# Your code here
drinks[(drinks['continent']== 'AS') & (drinks['total_litres_of_pure_alcohol'] > 5)] #Double specific condition with 'And'
```

## Part 6: Using .isin() for Multiple Values

These exercises practice the `.isin()` method for cleaner filtering.

### Exercise 22: Find North American and South American countries

Display all countries from North America (NA) and South America (SA).

**Hint:** Use `.isin(['NA', 'SA'])` on the continent column.

```python
# Your code here
drinks[drinks['continent'].isin(['NA', 'SA'])] # Multiple specific condition
```

### Exercise 23: Find specific countries

Display data for Spain, Portugal, and France.

**Hint:** Use `.isin()` on the 'country' column with a list of country names.

```python
# Your code here
drinks[drinks['country'].isin(['Spain','Portugal','France'])] # Specific multiple(3+) condition
```

### Exercise 24: Find countries NOT in Europe or Asia

Display countries that are NOT in Europe or Asia.

**Hint:** Use `~` (NOT operator) before `.isin()`: `~drinks['continent'].isin(['EU', 'AS'])`.

```python
# Your code here
drinks[~drinks['continent'].isin(['EU','As'])] #Wave symbol used for: 'Not'
    #Wave symbol
```

## Part 8: Finding Maximums and Minimums

These exercises practice finding extreme values.

### Exercise 29: Which country has the highest beer servings?

Find the country with the maximum beer servings.

**Hint:** Use `.loc[]` with `.idxmax()` or filter where `beer_servings == drinks['beer_servings'].max()`.

```python
# Your code here
drinks.loc[drinks['beer_servings'].idxmax(), 'country'] # idxmax() selects the max value of 'beer_servings', then displays country_name
```

```python
drinks[drinks['beer_servings'] == drinks['beer_servings'].max()] #Full row display of Max value
```

### Exercise 30: What is the maximum value for beer servings?

Display just the number (not the whole row).

**Hint:** Use `.max()` on the beer_servings column.

```python
# Your code here
drinks['beer_servings'].max() #Simple max value of beer servings
```

### Exercise 31: Which country has the highest total alcohol consumption?

Find the country with the maximum total litres of pure alcohol.

**Hint:** Use `.loc[drinks['total_litres_of_pure_alcohol'].idxmax()]`.

```python
# Your code here
drinks.loc[drinks['total_litres_of_pure_alcohol'].idxmax()]
```

### Exercise 32: Find the country with the lowest wine servings (excluding 0)

Find the country with the minimum wine servings, but only among countries that have wine servings > 0.

**Hint:** Filter first `drinks[drinks['wine_servings'] > 0]`, then find the minimum.

```python
# Your code here
drinks_with_no_wine = drinks[drinks['wine_servings'] > 0]   #Countries with more than 0 wine servings
drinks_with_no_wine = drinks_with_no_wine.loc[drinks_with_no_wine['wine_servings'].idxmin()] #Finding the minimum value of the group
drinks_with_no_wine #Printing
```
